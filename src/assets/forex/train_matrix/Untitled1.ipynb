```python
dates =pd.to_datetime(df['Date'])
```

```python
y = df['account_balance']
x = dates
```

```python
plt.rcParams['font.family'] = 'monospace'  # Change font family (e.g., 'serif', 'sans-serif', 'monospace')
plt.rcParams['font.size'] = 8              # Set global font size
plt.rcParams['font.weight'] = 'bold' 
```

```python
fig, ax = plt.subplots()
ax.plot(x, y, color='red', linestyle='-', linewidth=2, label='Connecting Line')
# If you'd like to add scatter points uncomment the line below
# ax.scatter(x, y)

# Set y-axis tick format
min_y = np.floor(min(y))
max_y = np.ceil(max(y))
ticks = np.arange(min_y, max_y, step=1500)  # Define a step value (e.g., 100)
ax.set_yticks(ticks)
ax.set_yticklabels([f'{tick:.2f}' for tick in ticks]) 

# Set x-axis to display dates in weekly intervals
ax.xaxis.set_major_locator(mdates.MonthLocator())  # Major ticks every month
ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m')) # Format for dates

# Rotate the date labels for better readability
plt.xticks(rotation=45)

# Add labels and title
ax.set_xlabel('Model Probability Threshold')
ax.set_ylabel('Account Balance ($)')
ax.set_title('Test Trades')

# Show the plot
plt.tight_layout()
plt.savefig('compundedprofit.png', transparent=True, dpi=300)
plt.show()



```
