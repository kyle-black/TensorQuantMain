```python
oswald_font = 'Oswald'
```

```python
plt.rcParams['font.family'] = 'monospace'  # Change font family (e.g., 'serif', 'sans-serif', 'monospace')
plt.rcParams['font.size'] = 12              # Set global font size
plt.rcParams['font.weight'] = 'bold' 
```

```python
x = list(proba_dict.keys())
y = list(proba_dict.values())

min_y = np.floor(min(y) * 100) / 100  # Get the minimum y value rounded down to hundredths
max_y = np.ceil(max(y) * 100) / 100 
```

```python
plt.plot(x, y, color='red', linestyle='-', linewidth=2, label='Connecting Line')
plt.scatter(x, y)



ticks = np.arange(min_y, max_y + 0.01, 0.01) 

plt.yticks(ticks, [f'{tick:.2f}' for tick in ticks]) 

# Add labels and title
plt.xlabel('Model Probability Threshold')
plt.ylabel('Test Accuracy')
plt.title('Test Accuracy at incremental probability thresholds')
ax = plt.gca()



# Show the plot
for tick in ax.get_yticks():
    tick_label = ax.yaxis.get_major_ticks()[list(ax.get_yticks()).index(tick)].label1
    if tick % 0.05 == 0:  # For ticks divisible by 0.05 (e.g., 0.50, 0.55, 0.60)
        tick_label.set_fontsize(8)
        tick_label.set_fontweight('bold')  # Make it bold
    else:
        tick_label.set_fontsize(10)  # Normal size for others

# Show the plot
plt.savefig('scatter_plot.png', transparent=True, dpi=300)
plt.show()

```
